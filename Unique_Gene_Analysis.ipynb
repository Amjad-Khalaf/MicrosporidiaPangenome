```python
# retrieve TE scores we computed for Fig 3
OrthogroupScores = {}
scoreFile = "orthogroup_TE_scores.txt"
with open(scoreFile) as file:
    for line in file:
        line = line.rstrip("\n")
        fields = line.split("\t")
        orthogroup = fields[0]
        score = float(fields[1])
        OrthogroupScores[orthogroup] = score
```

```python
# Importing the orthofinder results
import pandas as pd
orthogroupsTSV = "Orthogroups.GeneCount.tsv"
df = pd.read_csv(orthogroupsTSV, sep='\t')

orthogroupDict = {} #Key is species, values is list of orthogroups
speciesList = []
with open(orthogroupsTSV) as file:
    for line in file:
        line = line.rstrip("\n")
        fields = line.split("\t")
        if fields[0] == "Orthogroup":
            speciesList = fields.copy()
        else:
            for i, field in enumerate(fields):
                if field.startswith("OG"):
                    orthogroup = field
                else:
                    if int(field) > 0:
                        # retrieve position of field in list to find out what species it is
                        species = speciesList[i]
                        if species in orthogroupDict.keys() and OrthogroupScores[orthogroup] < 1:
                            orthogroupDict[species].append(orthogroup)
                        else:
                            if OrthogroupScores[orthogroup] < 1:
                                orthogroupDict[species] = [orthogroup]
```

```python
# this and the next code snippets apply to Ovavesiculida, but in principle, I ran the exact same thing for each order :)

# identify unique orthogroups
result = Ovavesiculida_core.difference(Mitosporidia_core, Metchnikovellida_core, Enterocytozoonida_core, Nosematida_core, Amblyosporida_core, Neopereziida_core, Glugeida_core, OrphanLineage_core)
print(len(result))
```

```python
# identify which of these unique orthogroups are widespread in their order (distribution >= 90%)

uniqueOrthos = []
for orthogroup in result:
    count = 0
    for genome in Ovavesiculida_genomes_sets:
        if orthogroup in genome:
            count = count + 1
    if count >= 0.9*len(Ovavesiculida_genomes_sets):
        uniqueOrthos.append(orthogroup)
        print(orthogroup)
```

```python
# map the identified orthogroups to the genes that constitute them

from pathlib import Path

GeneDict = {}
with open("Orthogroups.tsv") as file:
    GeneDict = {}

    for line in file:
        fields = line.rstrip("\n").split("\t")

        orthogroup = fields[0]

        if orthogroup in uniqueOrthos:
            genes = [gene for gene in fields[1:] if gene]
            GeneDict[orthogroup] = genes

for orthogroup, genes in GeneDict.items():
    print(orthogroup, genes)
```

```python
# get the relevant annotations per orthogroup now, based on the genes it includes

directory = Path("interproscan") # dir with my interproscan files

gene_to_orthogroup = {}

for orthogroup in uniqueOrthos:
    for gene in GeneDict[orthogroup]:
        gene_to_orthogroup[gene] = orthogroup

for file in directory.glob("*.interproscan"):
    with open(file) as f:
        for line in f:
            fields = line.rstrip("\n").split("\t")
            gene = fields[0]

            if gene in gene_to_orthogroup:
                orthogroup = gene_to_orthogroup[gene]
                updatedLine = ""
                for field in fields:
                    if "Reactome" in field:
                        pass
                    else:
                        updatedLine = updatedLine + "\t" + field
                print(orthogroup, file.name, updatedLine)
```
